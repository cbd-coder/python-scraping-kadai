```python
import requests
import pprint
from getpass import getpass

api_key = getpass('APIキーを入力してください:')

search_word = input('検索する駅を入力してください:')

url = 'https://maps.googleapis.com/maps/api/geocode/json'

response = requests.get(
    url,
    params={'key': api_key,
        'address': search_word}
)

json_data = response.json()

pprint.pprint(json_data)

latitute = json_data['results'][0]['geometry']['location']['lat']
longitute = json_data['results'][0]['geometry']['location']['lng']

print(latitute)
print(longitute)

restaurante_url = 'https://maps.googleapis.com/maps/api/place/nearbysearch/json'

response = requests.get(
    restaurante_url,
    params={
        'key': api_key,
        'location': f'{latitute},{longitute}',
        'radius': 500,
        'type': 'restaurant'
    }
)

restaurante_data = response.json()

pprint.pprint(restaurante_data)


for item in restaurante_data['results']:
        restaurante_name = item['name']
        restaurante_rating = item['rating']
        restaurante_address = item['vicinity']
        print(f'{restaurante_name}')
        print(f'{restaurante_rating}')
        print(f'{restaurante_address}')
```
